## Capstone Ideas

1) Tabular Data
- Use any common/known dataset (Kaggle Competitions, i.e. house pricing)
- Do EDA, pre-processing
- Structure your code cleanly with processing pipelines
- Do a classfication/regression model
- Compare different models such as RF, XGBoost, Neural Network
- Build a ML pipeline (using airflow, dagster) to train the model (store it somewhere)
- Build a score pipeline (offline batch inference pipeline)
- Deploy your model and do online inference on some samples (just deploy it locally or in a notebook settings)

2) Computer Vision
- Start with VGG16 model (or any base models)
- Tranfer learning/fine-tuning to classify a new set of images (classify your family photos)
- or fine-tune on a different dataset (CIFAR-10)
3) Computer Vision - Clustering using Embeddings
- Using VGG16 (or any other base model), use the model to produce embeddings for images
- For embeddings, you usually extract the last FC layer before the classification head.
- Clustering on these embeddings and explore interesting results.
https://www.kaggle.com/competitions/biotrove-clustering
- Try different embeddings from different CNNs and see which one works better for the competition above
- Contrast Learning: with the known species, you can fine-tune a CNN that classifies them well and then use the embedding for further classification.

4) Natural Language Processing
- Download a tokenizer base model from Hugging Face
- Download a dataset
- Train the model
- Do some form of evaluation
